## STEP 1 - Importing essential libraries:

```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
```

## 🪜 STEP 2 — Load Images

```python
# List of file paths for the images
image_paths = [
    "/kaggle/input/datasets/ahmedfakhar123/image-processing-samples/python.jpg",
    "/kaggle/input/datasets/ahmedfakhar123/image-processing-samples/numpy.png",
    "/kaggle/input/datasets/ahmedfakhar123/image-processing-samples/matplotlib.png"
]

# 1. Load images using PIL
# We use descriptive names to distinguish between PIL objects and NumPy arrays
python_img = Image.open(image_paths[0])
numpy_img = Image.open(image_paths[1])
matplotlib_img = Image.open(image_paths[2])

# 2. Pre-processing: Convert to RGB
# This ensures all images have 3 channels (Red, Green, Blue) 
# and removes Alpha (transparency) channels often found in PNGs
python_img_rgb = python_img.convert('RGB')
numpy_img_rgb = numpy_img.convert('RGB')
matplotlib_img_rgb = matplotlib_img.convert('RGB')

# 3. Convert PIL images to NumPy arrays for processing
python_arr = np.array(python_img_rgb)
numpy_arr = np.array(numpy_img_rgb)
matplotlib_arr = np.array(matplotlib_img_rgb)

# 4. Display image dimensions
# The shape will now consistently be (Height, Width, 3)
print("--- Image Dimensions (Height, Width, Channels) ---")
print(f"Python Logo:     {python_arr.shape}")
print(f"NumPy Logo:      {numpy_arr.shape}")
print(f"Matplotlib Logo: {matplotlib_arr.shape}")
```

### Showing Images:

```python
plt.figure(figsize=(20, 5))

plt.subplot(1, 3, 1)
plt.imshow(python_arr)
plt.title("Python")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(numpy_arr)
plt.title("NumPy")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(matplotlib_arr)
plt.title("Matplotlib")
plt.axis("off")

plt.show()
```

---

## 🌑 STEP 3 — Convert to Grayscale

### Concept:

Average RGB channels.

### Formula:

$$\text{Gray} = \frac{R + G + B}{3}$$

### Code:

```python
grayscale1 = python_arr.mean(axis=2)
grayscale2 = numpy_arr.mean(axis=2)
grayscale3 = matplotlib_arr.mean(axis=2)


plt.figure(figsize=(20, 5))

plt.subplot(1, 3, 1)
plt.imshow(grayscale1, cmap='gray')
plt.title("Python")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(grayscale2, cmap='gray')
plt.title("Numpy")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(grayscale3, cmap='gray')
plt.title("Matplotlib")
plt.axis("off")

plt.show()
```

## ☀️ STEP 3 — Brightness Adjustment

### Increase brightness:

```python
bright_python = np.clip(python_arr.astype(np.int32) + 50, 0, 255).astype(np.uint8)
bright_numpy = np.clip(numpy_arr.astype(np.int32) + 50, 0, 255).astype(np.uint8)
bright_matplotlib = np.clip(matplotlib_arr.astype(np.int32) + 50, 0, 255).astype(np.uint8)

plt.figure(figsize=(20, 5))

plt.subplot(1, 3, 1)
plt.imshow(bright_python)
plt.title("Python")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(bright_numpy)
plt.title("Numpy")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(bright_matplotlib)
plt.title("Matplotlib")
plt.axis("off")

plt.show()
```
