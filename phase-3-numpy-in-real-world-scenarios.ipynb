## ⚙️ Importing Required Libraries: NumPy & Matplotlib

```python
import numpy as np
import matplotlib.pyplot as plt
```

---

## 📊 Sample Dataset — Restaurant Sales Over Years 🍽️💰

Let’s work with a small but realistic dataset in NumPy to simulate **business sales analysis** 📈.

Each row represents a **restaurant**, and each column stores its **annual revenue**:

**Structure:**
`[restaurant_id, 2021, 2022, 2023, 2024, 2025]`

### 🔹 Dataset

```python
sales_data = np.array([
    [1, 150000, 175000, 140000, 210000, 300000],  # Spice Hub 🌶️
    [2, 120000, 140000, 155000, 180000, 260000],  # Urban Bites 🍔
    [3, 200000, 190000, 210000, 270000, 350000],  # Royal Dine 👑
    [4, 100000, 130000, 125000, 160000, 240000],  # Food Corner 🍕
    [5, 270000, 190000, 280000, 250000, 320000]   # Grill Master 🔥
])
```

---

## Resturant sales analysis
now we are analysing data of different resturants

### Shape of data
first we have to identify matrix/data shape

```python
print(f"Sales Data shape: {sales_data.shape}")
```

### Analysing data of first 3 resturants
We have to figure out what are the sales of 1st three resturants in five years

```python
print(f"\nSample data for 1st three resturants:\n {sales_data[:3]}")

# If we don't want resturant_id then
print(f"\nSample data for 1st three resturants:\n {sales_data[:, 1:]}")
```

### Total sales of each resturant
Now we come to know how to analyze data, now we are to find the total sales of each resturant over all years

```python
print(f"Total sales of 1st resturant: {np.sum(sales_data[0, 1:])}")
print(f"Total sales of 2st resturant: {np.sum(sales_data[1, 1:])}")
print(f"Total sales of 3st resturant: {np.sum(sales_data[2, 1:])}")
print(f"Total sales of 4st resturant: {np.sum(sales_data[3, 1:])}")
print(f"Total sales of 5st resturant: {np.sum(sales_data[4, 1:])}")
```

### Total sales per year
Now we are finding total sales of all resturants per year

```python
yearly_total = np.sum(sales_data[:, 1:], axis=0)
print(f"Yearly total sales: {yearly_total}")
```

### Overall sales of all resturants
We have to calculate all sales over five years of all resturants together

```python
overall_total = np.sum(yearly_total)
print(f"Overall total sales: {overall_total}")
```

### Minimum sales per resturant
Finding which is the minimum sales for each resturant

```python
min_sales = np.min(sales_data[:, 1:], axis=1)
print(f"Minimum sales per resturant: {min_sales}")
```

### Maximum sales per year
Finding the maximum sales in each year

```python
max_sales = np.max(sales_data[:, 1:], axis=0)
print(f"Maximum sales per year: {max_sales}")
```

### Average sales per resturant

```python
avg_sales = np.mean(sales_data[:, 1:], axis=1)
print(f"Average sales per resturant: {avg_sales}")
```

### Cumulative sales

```python
cumsum = np.cumsum(sales_data[:, 1:], axis=1)
print(f"Cumulative sales of all resturants:\n{cumsum}")
```

## Showing average cumulative sales on matplotlib graph

```python
plt.figure(figsize=(10, 8))
plt.plot(cumsum)
plt.title("Average")
plt.show()
```
