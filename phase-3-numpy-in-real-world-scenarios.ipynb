## ⚙️ Step 1: Import Required Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
```

### 📝 Explanation:
We use:

- **NumPy** for numerical computations
- **Matplotlib** for visualization

---

## 📊 Step 2: Creating the Dataset

We simulate a dataset representing **restaurant yearly sales**.

**Structure:**
`[restaurant_id, 2021, 2022, 2023, 2024, 2025]`

📝 Each row = one restaurant
📝 Each column = yearly revenue

<img src="https://framerusercontent.com/images/2NgQQXQrSdAtjHt9abwScZfyXI.svg?width=1713&height=951" width=800 />

```python
sales_data = np.array([
    [1, 150000, 175000, 140000, 210000, 300000],  # Spice Hub 🌶️
    [2, 120000, 140000, 155000, 180000, 260000],  # Urban Bites 🍔
    [3, 200000, 190000, 210000, 270000, 350000],  # Royal Dine 👑
    [4, 100000, 130000, 125000, 160000, 240000],  # Food Corner 🍕
    [5, 270000, 190000, 280000, 250000, 320000]   # Grill Master 🔥
])
```

---

## 🔍 Step 3: Understanding Data Shape

```python
print(f"Sales Data shape: {sales_data.shape}")
```

📝 **Explanation:**

* Rows → number of restaurants
* Columns → ID + yearly sales


---

## ✂️ Step 4: Data Slicing & Inspection
We have to figure out what are the sales of 1st three resturants in five years

```python
# First 3 restaurants (full data)
print(f"\nSample data for 1st three resturants:\n {sales_data[:3]}")

# Removing restaurant ID column
print(f"\nSample data for 1st three resturants:\n {sales_data[:, 1:]}")
```

---

## 💰 Step 5: Total Sales per Restaurant

```python
print(f"Total sales of 1st resturant: {np.sum(sales_data[0, 1:])}")
print(f"Total sales of 2st resturant: {np.sum(sales_data[1, 1:])}")
print(f"Total sales of 3st resturant: {np.sum(sales_data[2, 1:])}")
print(f"Total sales of 4st resturant: {np.sum(sales_data[3, 1:])}")
print(f"Total sales of 5st resturant: {np.sum(sales_data[4, 1:])}")
```

---

## 📆 Step 6: Total Sales per Year


```python
yearly_total = np.sum(sales_data[:, 1:], axis=0)
print(f"Yearly total sales: {yearly_total}")
```

📝 **Explanation:**

* `axis=0` → column-wise sum (per year)


---

## 🌍 Step 7: Overall Business Revenue

📝 Total revenue across all restaurants & years

```python
overall_total = np.sum(yearly_total)
print(f"Overall total sales: {overall_total}")
```

---

## 📉 Step 8: Minimum Sales per Restaurant
📝 Finds worst-performing year per restaurant

```python
min_sales = np.min(sales_data[:, 1:], axis=1)
print(f"Minimum sales per resturant: {min_sales}")
```

---

## 📈 Step 9: Maximum Sales per Year
📝 Finds best-performing restaurant each year

```python
max_sales = np.max(sales_data[:, 1:], axis=0)
print(f"Maximum sales per year: {max_sales}")
```

---

## ⚖️ Step 10: Average Sales per Restaurant
📝 Average yearly performance per restaurant

```python
avg_sales = np.mean(sales_data[:, 1:], axis=1)
print(f"Average sales per resturant: {avg_sales}")
```

---

## 🗓️ Step 11: Monthly Sales Estimation

📝 **Explanation:**

* Dividing yearly revenue into monthly estimate

```python
average_monthly = sales_data[:, 1:] / 12

# Rounded for better readability
print(np.round(average_monthly, 2))
```

---

## 🔄 Step 12: Cumulative Sales Growth

📝 **Explanation:**

* Shows growth trend over time
* Each value = total up to that year

<img src="https://community.fabric.microsoft.com/t5/image/serverpage/image-id/658578iFEC007F1621493C8?v=v2" width=600 />

```python
# Cumulative sum across years (axis=1 → row-wise)
cumsum = np.cumsum(sales_data[:, 1:], axis=1)
print(f"Cumulative sales of all resturants:\n{cumsum}")
```

---

## 📊 Step 13: Visualization — Growth Trend

📝 **Key Insight:**
This graph shows **overall business growth trend** 📈



```python
years = ["2021", "2022", "2023", "2024", "2025"]

plt.figure(figsize=(10, 8))

# Mean cumulative sales across all restaurants
plt.plot(np.mean(cumsum, axis=0))  
# axis=0 → averaging across restaurants

plt.xticks(range(len(years)), years)  # Set custom year labels
plt.ticklabel_format(style='plain', axis='y')  # Remove scientific notation

plt.title("Average Cumulative Sales Across Restaurants")
plt.xlabel("Years")
plt.ylabel("Sales")
plt.grid(True)

plt.show()
```
