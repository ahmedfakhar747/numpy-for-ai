# 🧰 1. Importing Core Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
```

📝 **What’s happening here?**

* `numpy` → handles image as numerical matrices (pixels)
* `matplotlib` → visualizes arrays as images


---

# 📊 2. Creating Base NumPy Arrays

```python
# 2D array (matrix-style data)
arr1 = np.array([[1, 2, 3, 4],
                 [5, 6, 7, 8]])

# Random floating values (useful in ML / noise simulation)
arr2 = np.random.rand(4, 5)

# Array full of ones (common in image masks)
arr3 = np.ones((3, 4))
```

🧠 **Insight:**

* Images are just **matrices of numbers**
* RGB images = 3D arrays
* Grayscale images = 2D arrays



---

# 🎨 3. Pixel Art Creation using NumPy (Image Engineering Mode)
Generate synthetic images by controlling pixel intensity and RGB channels using NumPy.

---

## 😃 3.1 Smiley Face (Grayscale Image)

<div align="center">
    
<img src="https://img.magnific.com/premium-vector/free-vector-pixel-style-emoticons_515427-66.jpg" width=500 />

</div> 


```python
# Create empty black canvas (grayscale image)
img1 = np.zeros((200, 200))

# 👀 Eyes (white pixels)
img1[50:70, 60:80] = 255
img1[50:70, 120:140] = 255

# 🙂 Mouth (white pixels block)
img1[120:150, 70:130] = 255
```


🧠 **What’s happening?**

* Each pixel = a number
* 0 = black, 255 = white
* We “draw” using slicing (super powerful concept)



---
## 🚗 3.2 Blue Car (RGB Image – 3 Channels)


🧠 **Key idea:**

* RGB = [Red, Green, Blue]
* Each pixel is a triplet
* Changing values = changing colors


```python
# RGB canvas (3 channels = Red, Green, Blue)
img2 = np.ones((300, 300, 3), dtype=np.uint8)

# 🚘 Car body (blue color)
img2[160:220, 50:250] = [0, 0, 255]

# 🏠 Roof of car
img2[120:160, 100:200] = [0, 0, 200]

# 🪟 Windows (light blue)
img2[125:150, 110:145] = [180, 220, 255]
img2[125:150, 155:190] = [180, 220, 255]

# 🛞 Wheels (dark gray)
img2[220:260, 80:110] = [50, 50, 50]
img2[220:260, 190:220] = [50, 50, 50]

# ⚙️ Wheel centers (inner part)
img2[230:250, 90:100] = [80, 80, 80]
img2[230:250, 200:210] = [80, 80, 80]

# 🛣️ Road line
img2[260:265, :] = [200, 200, 200]
```

---

# 💾 4. Saving NumPy Data as Files


🧠 **Why this matters:**

* `.npy` = fast NumPy binary format
* Used in ML pipelines for storing datasets efficiently


```python
# Saving arrays (non-image data)
np.save("array1.npy", arr1)
np.save("array2.npy", arr2)
np.save("array3.npy", arr3)

# Saving image arrays
np.save("image1.npy", img1)
np.save("image2.npy", img2)
```

---

# 📂 5. Loading Saved Arrays

🧠 **Concept:**

* You are basically restoring memory of arrays
* Super important for real ML workflows


```python
loaded_arr1 = np.load("array1.npy")
print("Loaded array1:\n", loaded_arr1)

loaded_arr2 = np.load("array2.npy")
print("Loaded array2:\n", loaded_arr2)

loaded_arr3 = np.load("array3.npy")
print("Loaded array3:\n", loaded_arr3)
```

---

# 🖼️ 6. Visualizing Arrays as Images

🧠 **Note:**

* `imshow()` converts matrix → visual image
* `axis('off')` removes clutter for clean visuals


```python
try:
    image1 = np.load("array1.npy")
    image2 = np.load("array2.npy")
    image3 = np.load("array3.npy")

    plt.figure(figsize=(15, 5))

    plt.subplot(1, 3, 1)
    plt.imshow(image1)
    plt.axis('off')
    plt.title("Array 1 Image")

    plt.subplot(1, 3, 2)
    plt.imshow(image2)
    plt.axis('off')
    plt.title("Array 2 Image")

    plt.subplot(1, 3, 3)
    plt.imshow(image3)
    plt.axis('off')
    plt.title("Array 3 Image")

    plt.show()

except FileNotFoundError:
    print("⚠️ File not found. Check saved .npy files.")
```

# 🚗 7. Displaying Pixel Art Images


```python
try:
    image4 = np.load("image1.npy")
    image5 = np.load("image2.npy")

    plt.figure(figsize=(15, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(image4)
    plt.axis('off')
    plt.title("Smiley Face")

    plt.subplot(1, 2, 2)
    plt.imshow(image5)
    plt.axis('off')
    plt.title("NumPy Car Art")

    plt.show()

except FileNotFoundError:
    print("⚠️ Image files missing!")
```


---

# 🔄 8. Image Inversion (Negative Effect)

🧠 **Logic:**

* Black → White
* White → Black
* Formula: `new_pixel = 255 - old_pixel`



```python
# Invert pixel values (classic image processing trick)
inverted_image4 = 255 - image4
inverted_image5 = 255 - image5
```

---

## 🎭 Display Inverted Images


```python
try:
    plt.figure(figsize=(15, 5))
    
    plt.subplot(1, 3, 1)
    plt.imshow(inverted_image4)
    plt.axis('off')
    plt.title("Smiley Face - Inverted")
    
    plt.subplot(1, 3, 2)
    plt.imshow(inverted_image5)
    plt.axis('off')
    plt.title("Blue Car - Inverted")
    plt.show()
except FileNotFoundError:
    print("The image file not found!")
```
