## IMporting necessary libraries

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Array creation

```python
arr1 = np.array([[1, 2, 3, 4], [5, 6, 7, 8]])
arr2 = np.random.rand(4, 5)
arr3 = np.ones((3, 4))
```

## Image array creation of pixels
To create a image with numpy arrays the following code is written. There are two images:

```python
# -------------------- IMAGE 1: SMILEY --------------------
# Creating a simple zero array
img1 = np.zeros((200, 200))

# eyes
img1[50:70, 60:80] = 255
img1[50:70, 120:140] = 255

# mouth
img1[120:150, 70:130] = 255

# -------------------- IMAGE 2: CAR --------------------
img2 = np.ones((300, 300, 3), dtype=np.uint8)  # black background

# ---------------- car body (blue) ----------------
img2[160:220, 50:250] = [0, 0, 255]

# car roof
img2[120:160, 100:200] = [0, 0, 200]

# windows
img2[125:150, 110:145] = [180, 220, 255]
img2[125:150, 155:190] = [180, 220, 255]

# ---------------- wheels ----------------
img2[220:260, 80:110] = [50, 50, 50]
img2[220:260, 190:220] = [50, 50, 50]

# wheel inner part
img2[230:250, 90:100] = [80, 80, 80]
img2[230:250, 200:210] = [80, 80, 80]

# ---------------- road line ----------------
img2[260:265, :] = [200, 200, 200]
```

## Saving array as image


```python
np.save("array1.npy", arr1)
np.save("array2.npy", arr2)
np.save("array3.npy", arr3)

np.save("image1.npy", img1)
np.save("image2.npy", img2)
```

## Loading the array
We can load image file as follows

```python
loaded_arr1 = np.load("array1.npy")
print(f"Loaded array1:\n{loaded_arr1}")

loaded_arr2 = np.load("array2.npy")
print(f"Loaded array2:\n{loaded_arr2}")

loaded_arr3 = np.load("array3.npy")
print(f"Loaded array3:\n{loaded_arr3}")
```

## Displaying array as an image using matplotlib

```python
try:
    # load images
    image1 = np.load("array1.npy")
    image2 = np.load("array2.npy")
    image3 = np.load("array3.npy")
    
    # single row layout
    plt.figure(figsize=(15, 5))
    
    plt.subplot(1, 3, 1)
    plt.imshow(image1)
    plt.axis('off')
    plt.title("Image 1")
    
    plt.subplot(1, 3, 2)
    plt.imshow(image2)
    plt.axis('off')
    plt.title("Image 2")
    
    plt.subplot(1, 3, 3)
    plt.imshow(image3)
    plt.axis('off')
    plt.title("Image 3")
    
    plt.show()
except FileNotFoundError:
    print("The image file not found!")
```

## Displaying images using matplotlib

```python
try:
    # load images
    image4 = np.load("image1.npy")
    image5 = np.load("image2.npy")
    
    # single row layout
    plt.figure(figsize=(15, 5))
    
    plt.subplot(1, 3, 1)
    plt.imshow(image4)
    plt.axis('off')
    plt.title("Smiley Face")
    
    plt.subplot(1, 3, 2)
    plt.imshow(image5)
    plt.axis('off')
    plt.title("Blue Car (NumPy Art)")
    plt.show()
except FileNotFoundError:
    print("The image file not found!")
```

## Inverting methods:
We can invert an image made using numpy arrays as...

```python
inverted_image4 = 255 - image4
inverted_image5 = 255 - image5

try:
    plt.figure(figsize=(15, 5))
    
    plt.subplot(1, 3, 1)
    plt.imshow(inverted_image4)
    plt.axis('off')
    plt.title("Smiley Face - Inverted")
    
    plt.subplot(1, 3, 2)
    plt.imshow(inverted_image5)
    plt.axis('off')
    plt.title("Blue Car - Inverted")
    plt.show()
except FileNotFoundError:
    print("The image file not found!")
```
